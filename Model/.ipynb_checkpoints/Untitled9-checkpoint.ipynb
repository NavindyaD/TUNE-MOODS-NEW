```python

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, BatchNormalization, Activation, Input
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping, ReduceLROnPlateau, CSVLogger
from tensorflow.keras.optimizers import Adam
import matplotlib.pyplot as plt
import numpy as np
import os
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
from tensorflow.keras.regularizers import l2 # For L2 regularization

# --- 1. Define Dataset Paths and Parameters ---
# Ensure these paths are correct for your system
train_dir = 'D:/Faces/train'
validation_dir = 'D:/Faces/validation'
img_size = (48, 48) # Standard for FER datasets, adjust if your images are larger/smaller
batch_size = 64
num_classes = 7 # angry, disgust, fear, happy, neutral, sad, surprise

# Create a directory for saving logs and plots
# Changed directory name to avoid overwriting previous runs and to ensure a fresh start
# If you encounter the OSError again, try changing 'model_output_v3' to something new like 'model_output_v4'
output_dir = 'model_output_v3'
os.makedirs(output_dir, exist_ok=True)

# --- 2. Data Loading and Preprocessing with Even More Aggressive Augmentation & Techniques ---
# Key for hitting higher accuracy: Robust Data Augmentation
train_datagen = ImageDataGenerator(
    rescale=1./255, # Normalize pixel values
    rotation_range=25, # Increased rotation
    width_shift_range=0.18, # Increased shift
    height_shift_range=0.18, # Increased shift
    shear_range=0.18, # Increased shear
    zoom_range=0.18, # Increased zoom
    horizontal_flip=True,
    fill_mode='nearest',
    brightness_range=[0.7, 1.3], # Added brightness augmentation
    channel_shift_range=50, # Added channel shift (adjust for grayscale, might need testing)
    # Consider adding Random Erasing/Cutout for further robustness
    # This requires custom implementation or using external libraries like Albumentations
    # For now, we'll stick to Keras's built-in options.
)

validation_datagen = ImageDataGenerator(rescale=1./255) # Only rescale for validation

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=img_size,
    color_mode='grayscale', # Assuming grayscale images
    batch_size=batch_size,
    class_mode='categorical',
    shuffle=True
)

validation_generator = validation_datagen.flow_from_directory(
    validation_dir,
    target_size=img_size,
    color_mode='grayscale', # Assuming grayscale images
    batch_size=batch_size,
    class_mode='categorical',
    shuffle=False # Keep order for evaluation
)

# --- 3. Model Architecture (Depper, Wider CNN with L2 Regularization) ---
# Aiming for a VGG-like structure with Batch Normalization and Dropout
# L2 regularization helps prevent overfitting by penalizing large weights
def build_advanced_cnn(input_shape, num_classes, l2_reg=0.0001):
    model = Sequential([
        # Block 1
        Conv2D(64, (3, 3), padding='same', kernel_regularizer=l2(l2_reg), input_shape=input_shape),
        BatchNormalization(),
        Activation('relu'),
        Conv2D(64, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        MaxPooling2D(pool_size=(2, 2)),
        Dropout(0.35), # Slightly increased dropout

        # Block 2
        Conv2D(128, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        Conv2D(128, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        MaxPooling2D(pool_size=(2, 2)),
        Dropout(0.35),

        # Block 3
        Conv2D(256, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        Conv2D(256, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        MaxPooling2D(pool_size=(2, 2)),
        Dropout(0.4), # Increased dropout

        # Block 4 (More capacity for complex features)
        Conv2D(512, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        Conv2D(512, (3, 3), padding='same', kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        MaxPooling2D(pool_size=(2, 2)),
        Dropout(0.4),

        Flatten(),
        Dense(512, kernel_regularizer=l2(l2_reg)),
        BatchNormalization(),
        Activation('relu'),
        Dropout(0.5), # Standard dropout for dense layers

        Dense(num_classes, activation='softmax')
    ])
    return model

model = build_advanced_cnn(input_shape=(img_size[0], img_size[1], 1), num_classes=num_classes, l2_reg=0.0001)

# --- 4. Compile the Model ---
# Very crucial: Optimal Learning Rate and Optimizer
# Using a slightly lower initial learning rate and more aggressive LR reduction might help.
# Also, consider a "warm-up" phase for the learning rate, but that's more advanced.
optimizer = Adam(learning_rate=0.0005) # Experiment with 0.001, 0.0005, 0.0001
model.compile(optimizer=optimizer,
              loss='categorical_crossentropy',
              metrics=['accuracy'])

model.summary()

# --- 5. Callbacks for Training (Fine-tuned for better convergence and avoiding overfitting) ---
# IMPORTANT: Changed the model saving format to .keras (native Keras format)
# This is the recommended format and can sometimes avoid HDF5-related locking issues.
checkpoint_filepath = os.path.join(output_dir, 'best_model_over_70.keras')
model_checkpoint_callback = ModelCheckpoint(
    filepath=checkpoint_filepath,
    save_weights_only=False, # Save the entire model
    monitor='val_accuracy',
    mode='max',
    save_best_only=True,
    verbose=1
)

early_stopping_callback = EarlyStopping(
    monitor='val_accuracy',
    patience=30, # Increased patience further to allow more time for LR reduction and convergence
    mode='max',
    restore_best_weights=True,
    verbose=1
)

reduce_lr_callback = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.2,
    patience=12, # Increased patience for LR reduction
    min_lr=0.0000001, # Even smaller min_lr
    mode='min',
    verbose=1
)

csv_logger_callback = CSVLogger(os.path.join(output_dir, 'training_log.csv'), append=True)

callbacks_list = [model_checkpoint_callback, early_stopping_callback, reduce_lr_callback, csv_logger_callback]

# --- 6. Train the Model ---
epochs = 200 # Increased epochs, EarlyStopping will handle stopping

history = model.fit(
    train_generator,
    steps_per_epoch=train_generator.samples // batch_size,
    epochs=epochs,
    validation_data=validation_generator,
    validation_steps=validation_generator.samples // batch_size,
    callbacks=callbacks_list
)

# --- 7. Evaluate the Model ---
print("\nEvaluating the best model on the validation set...")
# Load the best model using the .keras format
best_model = tf.keras.models.load_model(checkpoint_filepath)
val_loss, val_accuracy = best_model.evaluate(validation_generator)
print(f"Validation Loss (Best Model): {val_loss:.4f}")
print(f"Validation Accuracy (Best Model): {val_accuracy:.4f}")

# --- 8. Visualize Results ---
plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Validation'], loc='upper left')
plt.grid(True)

plt.tight_layout()
plt.savefig(os.path.join(output_dir, 'training_history.png'))
plt.show()

# --- 9. Further steps: Confusion Matrix, Classification Report for detailed analysis ---
print("\nGenerating Classification Report and Confusion Matrix...")
validation_generator.reset() # Reset validation generator to ensure correct order
y_true_classes = validation_generator.classes[validation_generator.index_array]

# Predict on the entire validation set
predictions = best_model.predict(validation_generator)
y_pred_classes = np.argmax(predictions, axis=1)

class_labels = list(train_generator.class_indices.keys())
print("\nClassification Report:")
print(classification_report(y_true_classes, y_pred_classes, target_names=class_labels))

print("\nConfusion Matrix:")
cm = confusion_matrix(y_true_classes, y_pred_classes)
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_labels, yticklabels=class_labels)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.savefig(os.path.join(output_dir, 'confusion_matrix.png'))
plt.show()

print(f"\nTraining completed. Best model saved to: {checkpoint_filepath}")
print(f"Training logs saved to: {os.path.join(output_dir, 'training_log.csv')}")
print(f"Training history plot saved to: {os.path.join(output_dir, 'training_history.png')}")
print(f"Confusion matrix plot saved to: {os.path.join(output_dir, 'confusion_matrix.png')}")
```

```python
# --- Save model as .h5 ---
h5_filepath = os.path.join(output_dir, 'best_model_over_70.h5')
best_model.save(h5_filepath)
print(f"Model also saved in HDF5 (.h5) format to: {h5_filepath}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Load the best trained model (saved as .keras)
model = tf.keras.models.load_model('model_output_v3/best_model_over_70.keras')

# Load validation data
val_path = "D:/Faces/validation"
val_aug = ImageDataGenerator(rescale=1./255)

val_data = val_aug.flow_from_directory(
    val_path,
    target_size=(48, 48),
    color_mode='grayscale',
    batch_size=32,
    class_mode='categorical',
    shuffle=False
)

# Predict on validation data
val_predictions = model.predict(val_data)
val_true_labels = val_data.classes
class_names = list(val_data.class_indices.keys())
n_classes = len(class_names)

# Binarize true labels for ROC
y_true = label_binarize(val_true_labels, classes=np.arange(n_classes))

# Plot ROC curves
plt.figure(figsize=(10, 8))
for i in range(n_classes):
    fpr, tpr, _ = roc_curve(y_true[:, i], val_predictions[:, i])
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, lw=2, label=f'{class_names[i]} (AUC = {roc_auc:.2f})')

# Diagonal reference line
plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve for Each Emotion Class')
plt.legend(loc='lower right')
plt.grid(True)
plt.tight_layout()
plt.savefig('model_output_v3/roc_curve.png')  # Save it
plt.show()
```
